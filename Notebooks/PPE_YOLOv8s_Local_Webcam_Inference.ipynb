```python
from pathlib import Path
from ultralytics import YOLO
import cv2

MODEL_PATH = Path.cwd() / "ppe_frozen.pt"
print("Working directory:", Path.cwd())
if not MODEL_PATH.is_file():
    raise FileNotFoundError(f"Model file not found: {MODEL_PATH}")

model = YOLO(str(MODEL_PATH))
print("Classes:", model.names)
```

```python
CAMERA_INDEX = 0
cap = cv2.VideoCapture(CAMERA_INDEX)

cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)

print(
    "Camera resolution:",
    int(cap.get(cv2.CAP_PROP_FRAME_WIDTH)),
    "x",
    int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT)),
)

if not cap.isOpened():
    cap.release()
    raise RuntimeError("Could not open the webcam. Try CAMERA_INDEX = 1.")

try:
    while True:
        ok, frame = cap.read()
        if not ok:
            print("Could not read a webcam frame.")
            break

        result = model.predict(frame, conf=0.05, imgsz=640, verbose=False)[0]
        cv2.imshow("PPE detection - press q to quit", result.plot())

        if cv2.waitKey(1) & 0xFF == ord("q"):
            break
finally:
    cap.release()
    cv2.destroyAllWindows()
```
